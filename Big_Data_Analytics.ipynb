# EX.NO: 2 — Install, configure and run Hadoop and HDFS  (verification program)
**Aim:** Install, configure and run Hadoop and HDFS.

Installing Hadoop is done by hand — follow `Experiment_02_Hadoop/INSTALL_HADOOP.md` (the manual's procedure).
This program then **verifies** the installation exactly as the manual's *Output* section does:
1. environment (JAVA_HOME, HADOOP_HOME, Path)  2. configuration files  3. running daemons (`jps`)
4. web UIs  5. an HDFS write / read / delete round trip.

It only reports what it really finds: if Hadoop is not installed or not running, those checks say **FAIL** —
nothing is simulated.

```python
import os
import shutil
import subprocess
import sys
import tempfile
import urllib.request
import xml.etree.ElementTree as ET
from pathlib import Path

sys.path.insert(0, ".")
from bda_common import finish

results = []


def record(status, description, detail=""):
    results.append((status, description, detail))
    print(f"[{status:^4}] {description}" + (f"  ->  {detail}" if detail else ""))


def run(cmd, timeout=60, **kw):
    """Run a command, return (returncode, combined output) or (None, error text)."""
    try:
        p = subprocess.run(cmd, capture_output=True, text=True, timeout=timeout,
                           shell=(os.name == "nt"), **kw)
        return p.returncode, (p.stdout + p.stderr).strip()
    except FileNotFoundError:
        return None, "command not found"
    except subprocess.TimeoutExpired:
        return None, f"timed out after {timeout}s"
```

## 1. Environment setup & verification (manual §3)

```python
def check_environment():
    java_home = os.environ.get("JAVA_HOME")
    hadoop_home = os.environ.get("HADOOP_HOME")
    record("PASS" if java_home and Path(java_home.strip('"')).exists() else "FAIL",
           "JAVA_HOME points to the JDK", java_home or "variable not set")
    record("PASS" if hadoop_home and Path(hadoop_home).exists() else "FAIL",
           "HADOOP_HOME points to the Hadoop folder", hadoop_home or "variable not set")

    path_entries = [p.strip().strip('"').rstrip("\\/").lower() for p in os.environ.get("PATH", "").split(os.pathsep)]
    for label, base in (("Java bin", java_home), ("Hadoop bin", hadoop_home and os.path.join(hadoop_home, "bin")),
                        ("Hadoop sbin", hadoop_home and os.path.join(hadoop_home, "sbin"))):
        if label == "Java bin" and base:
            base = os.path.join(base.strip('"'), "bin")
        ok = bool(base) and base.rstrip("\\/").lower() in path_entries
        record("PASS" if ok else "FAIL", f"{label} directory is on Path", base or "n/a")

    code, out = run(["java", "-version"])
    lines = [l for l in out.splitlines() if not l.startswith("Picked up")]
    out = "\n".join(lines)
    first = lines[0] if lines else ""
    record("PASS" if code == 0 else "FAIL", "java -version", first or out)
    if code == 0 and ('"1.8' not in out and "version \"8" not in out):
        record("INFO", "Java is not 1.8", "Hadoop 2.9.2 as in the manual needs Java 8; newer Java may fail")
```

## 2. Configuration files (manual §4.2)

```python
EXPECTED = {
    "core-site.xml": {"fs.defaultFS": "hdfs://localhost:9000"},
    "hdfs-site.xml": {"dfs.replication": "1"},
    "mapred-site.xml": {"mapreduce.framework.name": "yarn"},
    "yarn-site.xml": {"yarn.nodemanager.aux-services": "mapreduce_shuffle"},
}


def read_props(path):
    return {p.findtext("name", "").strip(): p.findtext("value", "").strip()
            for p in ET.parse(path).getroot().iter("property")}


def check_config():
    hadoop_home = os.environ.get("HADOOP_HOME")
    if not hadoop_home:
        record("SKIP", "configuration files", "HADOOP_HOME not set")
        return
    conf = Path(hadoop_home) / "etc" / "hadoop"
    for fname, props in EXPECTED.items():
        f = conf / fname
        if not f.exists():
            record("FAIL", f"{fname} exists", f"not found in {conf}")
            continue
        found = read_props(f)
        for key, want in props.items():
            got = found.get(key)
            record("PASS" if got == want else "FAIL", f"{fname}: {key} = {want}", f"found: {got}")
    hdfs = conf / "hdfs-site.xml"
    if hdfs.exists():
        found = read_props(hdfs)
        for key in ("dfs.namenode.name.dir", "dfs.datanode.data.dir"):
            val = found.get(key, "")
            local = val.replace("file:///", "").replace("file:/", "")
            record("PASS" if val and Path(local).exists() else "FAIL", f"hdfs-site.xml: {key} folder exists", val or "not set")
```

## 3. Hadoop version and running daemons (manual §5)

```python
DAEMONS = ("NameNode", "DataNode", "ResourceManager", "NodeManager")


def check_daemons():
    code, out = run(["hadoop", "version"])
    record("PASS" if code == 0 else "FAIL", "hadoop version", out.splitlines()[0] if out else "")
    code, out = run(["jps"])
    if code is None:
        record("FAIL", "jps (lists running Java daemons)", out)
        return
    print(out)
    running = {line.split()[1] for line in out.splitlines() if len(line.split()) > 1}
    for d in DAEMONS:
        record("PASS" if d in running else "FAIL", f"{d} is running (jps)")
```

## 4. Web UI verification (manual §6)

```python
UIS = [("NameNode UI", [50070, 9870]), ("ResourceManager UI", [8088]), ("DataNode UI", [50075, 9864])]


def check_web_uis():
    for name, ports in UIS:
        ok_port = None
        for port in ports:
            try:
                with urllib.request.urlopen(f"http://localhost:{port}", timeout=4) as r:
                    if r.status == 200:
                        ok_port = port
                        break
            except Exception:
                pass
        record("PASS" if ok_port else "FAIL", f"{name} answers on http://localhost:" + "|".join(map(str, ports)),
               f"port {ok_port}" if ok_port else "no response")
```

## 5. HDFS round trip (write → read → delete)

```python
def check_hdfs_roundtrip():
    code, _ = run(["hdfs", "dfs", "-ls", "/"], timeout=90)
    if code != 0:
        record("FAIL", "hdfs dfs -ls /", "HDFS is not reachable (is the NameNode running?)")
        return
    record("PASS", "hdfs dfs -ls /")
    with tempfile.TemporaryDirectory() as tmp:
        src = Path(tmp) / "hello.txt"
        src.write_text("hello hadoop\n")
        run(["hdfs", "dfs", "-rm", "-r", "-f", "/lab_ex02"])
        steps = [(["hdfs", "dfs", "-mkdir", "/lab_ex02"], "mkdir /lab_ex02"),
                 (["hdfs", "dfs", "-put", str(src), "/lab_ex02/hello.txt"], "put hello.txt"),
                 (["hdfs", "dfs", "-cat", "/lab_ex02/hello.txt"], "cat hello.txt")]
        content = ""
        for cmd, label in steps:
            code, out = run(cmd, timeout=90)
            if label.startswith("cat"):
                content = out
            record("PASS" if code == 0 else "FAIL", f"hdfs dfs {label}")
        record("PASS" if "hello hadoop" in content else "FAIL", "file content read back from HDFS equals what was written")
        code, _ = run(["hdfs", "dfs", "-rm", "-r", "-f", "/lab_ex02"])
        record("PASS" if code == 0 else "FAIL", "hdfs dfs -rm -r /lab_ex02 (clean up)")
```

## Run all checks and summarise

```python
def main():
    print("Hadoop / HDFS installation check\n" + "=" * 60)
    for title, fn in (("1. Environment", check_environment), ("2. Configuration", check_config),
                      ("3. Daemons", check_daemons), ("4. Web UIs", check_web_uis),
                      ("5. HDFS round trip", check_hdfs_roundtrip)):
        print(f"\n--- {title}")
        fn()
    failed = [r for r in results if r[0] == "FAIL"]
    passed = [r for r in results if r[0] == "PASS"]
    print("\n" + "=" * 60)
    print(f"SUMMARY: {len(passed)} passed, {len(failed)} failed")
    if failed:
        print("Hadoop is NOT fully installed / running.  Follow Experiment_02_Hadoop/INSTALL_HADOOP.md, "
              "start the daemons (start-all.cmd) and run this check again.")
    else:
        print("All checks passed: Hadoop and HDFS are installed, configured and running.")
    return 1 if failed else 0


if __name__ == "__main__":
    finish(main())
```

```python
import sys
from datetime import datetime, timezone
from pprint import pprint

sys.path.insert(0, ".")
from bda_common import get_mongo_db, finish, USING_MOCK


def now():
    return datetime.now(timezone.utc)
```

## Part 1 — Basic CRUD operations (Create, Read, Update, Delete)

```python
def crud_demo(db):
    db.student.drop()

    print("--- CREATE: db.student.insert({...})")
    db.student.insert_one({
        "regNo": "3014",
        "name": "Test Student",
        "course": {"courseName": "MCA", "duration": "3 Years"},
        "address": {"city": "Bangalore", "state": "KA", "country": "India"},
    })

    print("--- READ: db.student.find()")
    for doc in db.student.find():
        pprint(doc)

    print('--- READ: db.student.find({"regNo": "3014"})')
    for doc in db.student.find({"regNo": "3014"}):
        pprint(doc)

    print('--- UPDATE: db.student.update({"regNo": "3014"}, {$set: {"name": "Viraj"}})')
    res = db.student.update_one({"regNo": "3014"}, {"$set": {"name": "Viraj"}})
    print("matched:", res.matched_count, " modified:", res.modified_count)
    pprint(db.student.find_one({"regNo": "3014"}))

    print('--- DELETE: db.student.remove({"regNo": "3014"})')
    res = db.student.delete_one({"regNo": "3014"})
    print("deleted:", res.deleted_count)
    print("documents left in student collection:", db.student.count_documents({}))
```

## Part 2 — Working with arrays in MongoDB
Steps 1-7 of the manual.  The manual hard-codes an `ObjectId("5ec55af8…")`; that id only exists on the author's
machine, so the program uses the `_id` returned when the post is inserted.

```python
def arrays_demo(db):
    db.posts.drop()

    print("--- 2. Create and query a document")
    NEW_POST = {
        "name": "Working with Arrays",
        "user": "Database Rebel",
        "desc": "Maintaining an array of objects in a document",
        "content": "some content...",
        "created": now(),
        "updated": now(),
        "tags": ["mongodb", "arrays"],
    }
    post_id = db.posts.insert_one(NEW_POST).inserted_id
    pprint(db.posts.find_one())

    print("\n--- 3. Add an array element ($push)")
    NEW_COMMENT = {"user": "DB Learner",
                   "text": "Nice post, can I know more about the arrays in MongoDB?",
                   "updated": now()}
    db.posts.update_one({"_id": post_id}, {"$push": {"comments": NEW_COMMENT}})
    pprint(db.posts.find_one({"_id": post_id})["comments"])

    reply = {"user": "Database Rebel", "text": "Thank you, please look for updates", "updated": now()}
    db.posts.update_one({"_id": post_id}, {"$push": {"comments": reply}})
    print("after the second comment:")
    pprint(db.posts.find_one({"_id": post_id})["comments"])

    print("\n--- 4. Update an array element (positional operator $)")
    NEW_CONTENT = "Thank you, please look for updates - updated"
    db.posts.update_one({"_id": post_id, "comments.user": "Database Rebel"},
                        {"$set": {"comments.$.text": NEW_CONTENT}})
    pprint(db.posts.find_one({"_id": post_id})["comments"][1])

    print("\n--- 5. Delete an array element ($pull)")
    db.posts.update_one({"_id": post_id}, {"$pull": {"comments": {"user": "Database Rebel"}}})
    pprint(db.posts.find_one({"_id": post_id})["comments"])

    print("\n--- 6. Add a new field to all objects in the array ($[])")
    try:
        db.posts.update_one({"_id": post_id}, {"$set": {"comments.$[].likes": 0}})
    except Exception as exc:
        if not USING_MOCK:
            raise
        print("(test backend mongomock does not implement the $[] operator - step NOT verified:", type(exc).__name__, ")")
        return
    pprint(db.posts.find_one({"_id": post_id})["comments"])

    print("\n--- 7. Update a specific array element based on a condition (arrayFilters)")
    NEW_COMMENT = {"user": "DB Learner", "text": "Thanks for the updates!", "updated": now()}
    db.posts.update_one({"_id": post_id}, {"$push": {"comments": NEW_COMMENT}})
    db.posts.update_one(
        {"_id": post_id},
        {"$inc": {"comments.$[ele].likes": 1}},
        array_filters=[{"ele.user": "DB Learner", "ele.likes": {"$exists": False}}],
    )
    pprint(db.posts.find_one({"_id": post_id})["comments"])
```

```python
def main():
    db = get_mongo_db("lab_ex04")
    if db is None:
        return 2
    crud_demo(db)
    print("\n" + "=" * 70 + "\nWORKING WITH ARRAYS\n" + "=" * 70)
    arrays_demo(db)
    return 0


if __name__ == "__main__":
    finish(main())
```

# EX.NO: 5 — Functions: Count, Sort, Limit, Skip, Aggregate using MongoDB
**Aim:** Implement the MongoDB functions Count – Sort – Limit – Skip – Aggregate.

**Needs a running MongoDB server** (see README → *MongoDB*).  The same commands in the Mongo shell are in
`Experiment_05.js`.  Data: the 8 sample products printed in the manual.

```python
import sys
from pprint import pprint

sys.path.insert(0, ".")
from bda_common import get_mongo_db, finish

PRODUCTS = [
    {"name": "Laptop",       "category": "Electronics", "price": 1200, "stock": 50,  "rating": 4.5},
    {"name": "Mouse",        "category": "Electronics", "price": 25,   "stock": 200, "rating": 4.0},
    {"name": "Keyboard",     "category": "Electronics", "price": 75,   "stock": 150, "rating": 4.2},
    {"name": "Desk Chair",   "category": "Furniture",   "price": 150,  "stock": 75,  "rating": 4.8},
    {"name": "Monitor",      "category": "Electronics", "price": 300,  "stock": 100, "rating": 4.7},
    {"name": "Coffee Table", "category": "Furniture",   "price": 100,  "stock": 40,  "rating": 3.9},
    {"name": "External HDD", "category": "Storage",     "price": 80,   "stock": 120, "rating": 4.1},
    {"name": "Webcam",       "category": "Electronics", "price": 60,   "stock": 90,  "rating": 4.6},
]


def show(title, cursor):
    print(f"\n--- {title}")
    for d in cursor:
        d.pop("_id", None)
        print(d)
```

```python
def run_experiment(db):
    db.products.drop()
    db.products.insert_many([dict(p) for p in PRODUCTS])
    print("Inserted", db.products.count_documents({}), "documents into store.products")

    print("\n--- db.products.countDocuments({})")
    print(db.products.count_documents({}))
    print('\n--- db.products.countDocuments({ "category": "Electronics" })')
    print(db.products.count_documents({"category": "Electronics"}))

    show('db.products.find().sort({ "price": 1 })', db.products.find().sort("price", 1))
    show('db.products.find({ "category": "Electronics" }).sort({ "rating": -1 })',
         db.products.find({"category": "Electronics"}).sort("rating", -1))

    show('db.products.find().sort({ "price": -1 }).limit(3)', db.products.find().sort("price", -1).limit(3))

    show('db.products.find().sort({ "name": 1 }).skip(2).limit(3)',
         db.products.find().sort("name", 1).skip(2).limit(3))

    print("\n--- db.products.aggregate([ {$group: ...}, {$sort: ...} ])")
    pipeline = [
        {"$group": {"_id": "$category",
                    "averagePrice": {"$avg": "$price"},
                    "totalStock": {"$sum": "$stock"}}},
        {"$sort": {"averagePrice": -1}},
    ]
    for d in db.products.aggregate(pipeline):
        print(d)


def main():
    db = get_mongo_db("store")
    if db is None:
        return 2
    run_experiment(db)
    return 0


if __name__ == "__main__":
    finish(main())
```

```python
import re
import subprocess
import sys
from collections import defaultdict
from pathlib import Path

sys.path.insert(0, ".")
from bda_common import DATA_DIR, BASE_DIR, finish
```

## Part A — Mapper, Shuffle/Sort, Reducer in Python

```python
with open(DATA_DIR / "data.txt", "r") as f:
    lines = f.readlines()

mapper_output = []
for line in lines:
    line = line.strip().lower()
    line = re.sub(r"[^\w\s']", "", line)
    words = re.split(r"\s+", line)
    for word in words:
        if word:
            mapper_output.append((word, 1))

print("=== Mapper Output ===")
for word, count in mapper_output:
    print(f"{word}\t{count}")

sorted_output = sorted(mapper_output, key=lambda x: x[0])
print("\n=== Sorted Output ===")
for word, count in sorted_output:
    print(f"{word}\t{count}")

reducer_output = defaultdict(int)
for word, count in sorted_output:
    reducer_output[word] += count

print("\n=== Reducer Output ===")
for word, count in reducer_output.items():
    print(f"{word}\t{count}")
```

## Part B — the streaming pipeline  `cat data.txt | ./mapper.py | sort | ./reducer.py`
On Linux/macOS the manual runs this in a shell (`chmod +x mapper.py reducer.py`, then the pipe).  To work on
Windows too, the same three stages are chained here with `subprocess`; `sorted()` plays the role of `sort`.

```python
def run_pipeline():
    py = sys.executable
    mapped = subprocess.run([py, str(BASE_DIR / "mapper.py")], input=(DATA_DIR / "data.txt").read_text(),
                            capture_output=True, text=True, check=True).stdout
    shuffled = "".join(sorted(mapped.splitlines(keepends=True)))
    reduced = subprocess.run([py, str(BASE_DIR / "reducer.py")], input=shuffled,
                             capture_output=True, text=True, check=True).stdout
    return reduced


pipeline_output = run_pipeline()
print("=== Output of  cat data.txt | mapper.py | sort | reducer.py ===")
print(pipeline_output)

pipeline_counts = {w: int(c) for w, c in (l.split("\t") for l in pipeline_output.splitlines())}
print("Part A and Part B give identical counts:", pipeline_counts == dict(reducer_output))
```

# EX.NO: 7 — A MapReduce program that processes a dataset
**Aim:** Implement a MapReduce program that processes a dataset.

**Dataset:** `datasets/data.csv` — UCI *Wine Quality* (red wine, 1599 rows, `;`-separated).
The mapper emits `(quality, volatile acidity)` for each row, the data is split into 4 slices (4 "mappers"),
shuffled by quality class into `shuffled.pkl`, and the reducer prints the **average volatile acidity per quality class**.

```python
import pickle
import sys

import pandas as pd

sys.path.insert(0, ".")
from bda_common import DATA_DIR, OUT_DIR

SHUFFLED_PKL = OUT_DIR / "shuffled.pkl"
```

## Mapper program

```python
data = pd.read_csv(DATA_DIR / "data.csv", sep=";")

slice1 = data.iloc[0:399, :]
slice2 = data.iloc[400:800, :]
slice3 = data.iloc[801:1200, :]
slice4 = data.iloc[1201:, :]


def mapper(data):
    mapped = []
    for index, row in data.iterrows():
        mapped.append((row['quality'], row['volatile acidity']))
    return mapped


map1 = mapper(slice1)
map2 = mapper(slice2)
map3 = mapper(slice3)
map4 = mapper(slice4)

shuffled = {
    3.0: [],
    4.0: [],
    5.0: [],
    6.0: [],
    7.0: [],
    8.0: [],
}
for i in [map1, map2, map3, map4]:
    for j in i:
        shuffled[j[0]].append(j[1])

with open(SHUFFLED_PKL, 'wb') as file:
    pickle.dump(shuffled, file)

print("Data has been mapped. Now, run the reducer to reduce the contents in shuffled.pkl file.")
```

## Reducer program

```python
with open(SHUFFLED_PKL, 'rb') as file:
    shuffled = pickle.load(file)


def reduce(shuffled_dict):
    reduced = {}
    for i in shuffled_dict:
        reduced[i] = sum(shuffled_dict[i]) / len(shuffled_dict[i])
    return reduced


final = reduce(shuffled)
print("Average volatile acidity in different classes of wine: ")
for i in final:
    print(i, ':', final[i])
```

# EX.NO: 8 — Clustering techniques using Spark
**Aim:** Implement clustering techniques using Spark (K-Means with PySpark MLlib).

**Needs:** PySpark (`pip install pyspark`) and a Java runtime (JDK 17 or newer for PySpark 4.x, JDK 8/11/17 for
PySpark 3.x).  Spark runs in *local mode* inside this process — no cluster or service has to be started.
**Data:** generated in the program: 100 points around (10, 10) and 100 points around (50, 50).

```python
import random
import sys

from pyspark.sql import SparkSession
from pyspark.ml.clustering import KMeans
from pyspark.ml.feature import VectorAssembler
```

## 1. Setup and data generation

```python
import os
os.environ.setdefault("PYSPARK_PYTHON", sys.executable)
spark = (SparkSession.builder
         .appName("ClusteringExperiment")
         .master("local[*]")
         .config("spark.ui.showConsoleProgress", "false")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
print("Spark version:", spark.version)

random.seed(42)

data_points = []
for _ in range(100):
    data_points.append((10 + random.uniform(-2, 2), 10 + random.uniform(-2, 2)))
for _ in range(100):
    data_points.append((50 + random.uniform(-2, 2), 50 + random.uniform(-2, 2)))

df = spark.createDataFrame(data_points, ["feature1", "feature2"])

print("Generated Data:")
df.show(5)
```

## 2. Data preparation

```python
assembler = VectorAssembler(inputCols=["feature1", "feature2"], outputCol="features")

transformed_df = assembler.transform(df)

print("Transformed Data with 'features' column:")
transformed_df.show(5)
```

## 3. Model training and prediction
K-Means with k = 2, as that is how the data was generated.

```python
kmeans = KMeans(k=2, seed=1)

model = kmeans.fit(transformed_df)

predictions = model.transform(transformed_df)

print("Predictions with 'prediction' (cluster) column:")
predictions.show(5)

cost = model.summary.trainingCost
print(f"Within Set Sum of Squared Errors (WSSSE): {cost}")

centers = model.clusterCenters()
print("Cluster Centers:")
for center in centers:
    print(center)

print("\nPoints per cluster:")
predictions.groupBy("prediction").count().orderBy("prediction").show()
```

```python
spark.stop()
```

# EX.NO: 9 — An application that stores big data in MongoDB using Hadoop
**Aim:** Implement an application that stores big data in MongoDB / Pig using Hadoop / R.

The manual's design: a **mapper** reads a large CSV, turns every record into a JSON/BSON document, and the
**Hadoop–MongoDB connector** writes the documents to a MongoDB collection (`bigdata_db.transactions`).

What this program does for real:
1. runs the manual's mapper (`Experiment_09_mapper.py`) over `datasets/transactions.csv`;
2. stores the resulting documents in MongoDB (**needs a running MongoDB server**) and verifies them like the
   Mongo shell would (count + sample documents).

What it does **not** do: submit the job to a Hadoop cluster.  That command is in
`Experiment_09_hadoop_streaming.sh` (unverified — see the note at its top and in the README).
The manual's "Output" (job counters for 1,000,000 records) is an illustrative log and cannot be reproduced;
`transactions.csv` is a 10-row sample (rows 1-2 are the ones printed in the manual).

```python
import json
import shutil
import subprocess
import sys

sys.path.insert(0, ".")
from bda_common import BASE_DIR, DATA_DIR, get_mongo_db, finish
```

## Step 1 — environment: which parts of the Hadoop + MongoDB stack are present?

```python
print("hadoop on PATH :", shutil.which("hadoop") or "NOT FOUND  (the Hadoop job in the .sh file cannot be run here)")
print("hdfs   on PATH :", shutil.which("hdfs") or "NOT FOUND")
```

## Step 2 — run the mapper (map phase) over the CSV

```python
def run_mapper():
    csv_text = (DATA_DIR / "transactions.csv").read_text()
    out = subprocess.run([sys.executable, str(BASE_DIR / "Experiment_09_mapper.py")], input=csv_text,
                         capture_output=True, text=True, check=True).stdout
    print("Mapper output (key<TAB>json), first 3 lines:")
    for line in out.splitlines()[:3]:
        print(" ", line)
    return [json.loads(line.split("\t", 1)[1]) for line in out.splitlines()]


documents = run_mapper()
print(f"\nMapper produced {len(documents)} documents")
```

## Step 3 — store the documents in MongoDB and verify

```python
def store_and_verify(docs):
    db = get_mongo_db("bigdata_db")
    if db is None:
        print("The documents above were NOT stored, because there is no MongoDB server to store them in.")
        return 2
    db.transactions.drop()
    result = db.transactions.insert_many([dict(d) for d in docs])
    print(f"Inserted {len(result.inserted_ids)} documents into bigdata_db.transactions")

    print("db.transactions.countDocuments({})  ->", db.transactions.count_documents({}))
    print("db.transactions.find().limit(3):")
    for d in db.transactions.find().limit(3):
        d.pop("_id")
        print(" ", d)
    total = list(db.transactions.aggregate([{"$group": {"_id": None, "revenue": {"$sum": {"$multiply": ["$quantity", "$price"]}}}}]))
    print("total revenue (quantity x price):", round(total[0]["revenue"], 2))
    return 0


if __name__ == "__main__":
    finish(store_and_verify(documents))
```
